## 4. Student Risk Analysis

Student risk analysis is used to identify students who may require additional academic support.

A rule-based risk classification is created using attendance, final percentage, and backlogs.

The risk classification is intended as an analytical indicator and not as a definitive prediction of student outcomes.

```python
def calculate_risk(row):
    risk_score = 0

    if row["Attendance"] < 60:
        risk_score += 2
    elif row["Attendance"] < 75:
        risk_score += 1

    if row["Final_Percentage"] < 50:
        risk_score += 2
    elif row["Final_Percentage"] < 65:
        risk_score += 1

    if row["Backlogs"] >= 3:
        risk_score += 2
    elif row["Backlogs"] >= 1:
        risk_score += 1

    if risk_score >= 4:
        return "High Risk"
    elif risk_score >= 2:
        return "Medium Risk"
    else:
        return "Low Risk"
```

```python
df["Risk_Level"] = df.apply(calculate_risk, axis=1)
```

```python
df["Risk_Level"].value_counts()
```

# What-If Performance Analysis

The What-If Analysis provides an analytical simulation of how changes in selected academic factors could affect a student's estimated performance.

The simulation is based on relationships observed in the dataset and should not be interpreted as a guaranteed prediction.

```python
current_percentage = student["Final_Percentage"]
current_attendance = student["Attendance"]
current_study_hours = student["Study_Hours_Per_Day"]
print("Current Final Percentage:", current_percentage)
print("Current Attendance:", current_attendance)
print("Current Study Hours:", current_study_hours)
```

```python
improved_attendance = min(current_attendance + 10, 100)
improved_study_hours = current_study_hours + 1

attendance_change = improved_attendance - current_attendance
study_change = improved_study_hours - current_study_hours

attendance_effect = attendance_change * 0.10
study_effect = study_change * 1.5

estimated_percentage = min(
    current_percentage + attendance_effect + study_effect,
    100
)

print("Current Percentage:", round(current_percentage, 2))
print("Improved Attendance:", round(improved_attendance, 2))
print("Improved Study Hours:", round(improved_study_hours, 2))
print("Estimated Percentage:", round(estimated_percentage, 2))
```
